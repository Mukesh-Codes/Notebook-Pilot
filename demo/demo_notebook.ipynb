# Small regression experiment
This notebook generates noisy linear data, fits a model, and plots diagnostics.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Generate synthetic data for the regression task
rng = np.random.default_rng(7)
x = np.linspace(-3, 3, 80)
y = 2.4 * x - 0.7 + rng.normal(0, 0.8, size=x.shape)
print(f'Generated {len(x)} samples')
```

## Fit a linear model

```python
# Fit a line and report the learned coefficients
slope, intercept = np.polyfit(x, y, deg=1)
pred = slope * x + intercept
mse = np.mean((pred - y) ** 2)
print(f'slope={slope:.3f}, intercept={intercept:.3f}, mse={mse:.3f}')
```

```python
plt.scatter(x, y, s=15, alpha=0.6)
plt.plot(x, pred)
plt.title('Linear fit')
plt.show()
```

## Unrelated experiment

```python
random_matrix = rng.normal(size=(300, 300))
singular_values = np.linalg.svd(random_matrix, compute_uv=False)
print('Largest singular value:', singular_values[0])
```
